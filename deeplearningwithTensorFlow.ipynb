## What are Tensors?

In simple terms, tensors are the fundamental data structures used in deep learning. WE can think of them as multi-dimensional arrays or a generalization of vectors and matrices:

- A 0D tensor is a single number (scalar)
- 1D tensor is a list of numbers (vectors)
- A 2D tensor is a table of numbers (matrix)

## Manipulating Tensors

**Basic Operations** 

`+` `-` `x` `/`

```python
# Addidng Values
tensor = tf.constant([[2, 6], [2, 5]])
tensor
```

```python
tensor_new = tensor + 10
tensor_new
```

```python
# Multiplication
tensor_mult = tensor * 10
tensor_mult
```

```python
# using the built-in function
tensor_new = tf.multiply(tensor, 100)
tensor_new
```

```python
tensor_new = tf.divide(tensor, 100)
tensor_new
```

**Matrix Multiplication**

**read** = https://www.mathsisfun.com/algebra/matrix-multiplying.html

**Try MM** = http://matrixmultiplication.xyz/


```python
tensor_1 = tf.constant([[[1, 2, 1], [0, 1, 0], [2, 3, 4]]])
tensor_1
```

```python
tensor_2 = tf.constant([[2, 5], [6, 7], [1, 8]])
tensor_2
```

```python
tf.matmul(tensor_1, tensor_2)
```

```python
# built-in python mehtod for matrix multiplication
tensor_1 @ tensor_2
```

**For Matrix multiplication, the inner dimensions should match, for eg, we can only multiply matrices of dimension 5 x 4 and 4 x 6 as 4=4 here, and the new matrix thus produced will be 5 x 6**

```python
# Transpose
transposed_tensor = tf.transpose(tensor_1)
transposed_tensor
```
